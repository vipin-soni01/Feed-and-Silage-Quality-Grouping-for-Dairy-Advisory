```python
features = [
    "dm.s",
    "ash.s",
    "cp.s",
    "ee.s",
    "ndf.s",
    "adf.s",
    "starch.s",
    "pH",
    "ammonia.s",
    "glucose.s",
    "fructose.s",
    "ethanol.s",
    "lactic.ac.s",
    "acetic.ac.s",
    "propionic.ac.s",
    "butyric.ac.s",
    "dm.loss"
]
```

```python
#  Qualty Refrence Groups

data["quality"] = pd.cut(
    data["fqi"],
    bins=[-float("inf"), 47.7, 57.4, float("inf")],
    labels=["Lower", "Good", "Excellent"]
)
```

```python
data["quality"].value_counts()
```

```python
data[["fqi", "quality"]].head(20)
```

```python
from sklearn.metrics import balanced_accuracy_score
import numpy as np

baseline_data = data[["moisture", "fqi"]].dropna().copy()

# True quality: 1 = Excellent, 0 = Not Excellent
y_true = (baseline_data["fqi"] >= 57.4).astype(int)

thresholds = np.sort(baseline_data["moisture"].unique())

results = []

for threshold in thresholds:
    y_pred = (baseline_data["moisture"] >= threshold).astype(int)

    score = balanced_accuracy_score(y_true, y_pred)

    results.append({
        "threshold": threshold,
        "balanced_accuracy": score
    })

threshold_results = pd.DataFrame(results)

best_result = threshold_results.loc[
    threshold_results["balanced_accuracy"].idxmax()
]

best_result
```

```python
best_threshold = best_result["threshold"]

print("Best moisture threshold:", best_threshold)
print("Balanced accuracy:", best_result["balanced_accuracy"])
```

```python
baseline_data["predicted"] = (
    baseline_data["moisture"] >= best_threshold
).astype(int)

print(
    "Baseline balanced accuracy:",
    balanced_accuracy_score(
        y_true,
        baseline_data["predicted"]
    )
)
```

```python
#  K Means using Elbow method  , Silhouette score

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

inertia = []
silhouette_scores = []

k_values = range(2, 7)

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_scaled)
    
    inertia.append(kmeans.inertia_)
    silhouette_scores.append(
        silhouette_score(X_scaled, labels)
    )
```

```python
results_k = pd.DataFrame({
    "k": list(k_values),
    "inertia": inertia,
    "silhouette": silhouette_scores
})

results_k
```

```python
k_values = range(2, 11)

inertia = []
silhouette_scores = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_scaled)
    
    inertia.append(kmeans.inertia_)
    silhouette_scores.append(
        silhouette_score(X_scaled, labels)
    )

results_k = pd.DataFrame({
    "k": list(k_values),
    "inertia": inertia,
    "silhouette": silhouette_scores
})

results_k
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    results_k["k"],
    results_k["inertia"],
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.xticks(range(2, 11))
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    results_k["k"],
    results_k["silhouette"],
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score vs Number of Clusters")
plt.xticks(range(2, 11))
plt.show()
```

```python
fermentation_features = [
    "pH",
    "ammonia.s",
    "glucose.s",
    "fructose.s",
    "ethanol.s",
    "lactic.ac.s",
    "acetic.ac.s",
    "propionic.ac.s",
    "butyric.ac.s",
    "dm.loss"
]

X_fermentation = data[fermentation_features].copy()
```

```python
imputer_fermentation = SimpleImputer(strategy="median")

X_fermentation_imputed = imputer_fermentation.fit_transform(
    X_fermentation
)

scaler_fermentation = StandardScaler()

X_fermentation_scaled = scaler_fermentation.fit_transform(
    X_fermentation_imputed
)
```

```python
k_values = range(2, 11)

fermentation_inertia = []
fermentation_silhouette = []

for k in k_values:
    kmeans_temp = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels_temp = kmeans_temp.fit_predict(
        X_fermentation_scaled
    )

    fermentation_inertia.append(
        kmeans_temp.inertia_
    )

    fermentation_silhouette.append(
        silhouette_score(
            X_fermentation_scaled,
            labels_temp
        )
    )

fermentation_results = pd.DataFrame({
    "k": list(k_values),
    "inertia": fermentation_inertia,
    "silhouette": fermentation_silhouette
})

fermentation_results
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    fermentation_results["k"],
    fermentation_results["silhouette"],
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Fermentation Features - Silhouette Score")
plt.xticks(range(2, 11))
plt.show()
```

```python
kmeans_fermentation = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

labels_fermentation = kmeans_fermentation.fit_predict(
    X_fermentation_scaled
)
```

```python
data["cluster_fermentation"] = labels_fermentation
```

```python
data["cluster_fermentation"].value_counts().sort_index()
```

```python
evaluation_fermentation = data[
    data["quality"].notna()
].copy()

table_fermentation = pd.crosstab(
    evaluation_fermentation["cluster_fermentation"],
    evaluation_fermentation["quality"]
)

table_fermentation
```

```python
purity_fermentation = (
    table_fermentation.max(axis=1).sum()
    / table_fermentation.values.sum()
)

print("Fermentation K=5 Purity:", purity_fermentation)
```

```python
ari_fermentation = adjusted_rand_score(
    evaluation_fermentation["quality"],
    evaluation_fermentation["cluster_fermentation"]
)

print("Fermentation K=5 ARI:", ari_fermentation)
```

```python
silhouette_fermentation = silhouette_score(
    X_fermentation_scaled,
    labels_fermentation
)

print(
    "Fermentation K=5 Silhouette:",
    silhouette_fermentation
)
```
